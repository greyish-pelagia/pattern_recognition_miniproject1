# Preprocess one image

Run from the project root. Displays the original image, then the restored image.


```python
from pathlib import Path
from PIL import Image
from IPython.display import display
from src.enhancement import restore_images
```

```python
image_path = next(p for p in sorted(Path("data").rglob("*")) if p.suffix.lower() in {".jpg", ".jpeg", ".png"})
with Image.open(image_path) as source:
    image = source.convert("RGB")
```

```python
restored = restore_images(
    [image],
    sdxl_checkpoint="models/sd_xl_base_1.0_0.9vae.safetensors",
    supir_checkpoint="models/SUPIR-v0Q.ckpt",
    device="mps",
    dry_run=False,
)[0]
display(image, restored)
```
